```python
from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter



text_splitter = RecursiveCharacterTextSplitter(
    chunk_size = 300,        # 每個片段大約 300 字
    chunk_overlap  = 30,     # 片段間重疊 30 字
    add_start_index = True,  # 記住每個片段在原文中的位置
)

loader = TextLoader("doc.md", encoding="utf-8")
data = loader.load()

recursive_splits = text_splitter.split_documents(data) 
chunks = [doc.page_content for doc in recursive_splits]

for i, chunk in enumerate(chunks):
    
    print(f"[{i}] {chunk}\n")
```

```python
from sentence_transformers import SentenceTransformer
embedding_model = SentenceTransformer("Qwen/Qwen3-Embedding-0.6B", trust_remote_code=True)
def get_embedding(text, is_query=False):
    prefix = "query: " if is_query else "passage: "
    return embedding_model.encode([prefix + text])[0].tolist()

test_embedding=get_embedding("測試")
print(len(test_embedding))
print(test_embedding)
```

```python
embeddings=[get_embedding(chunk) for chunk in chunks]
print(len(embeddings))
print(embeddings[0])
```

```python
import chromadb

chromadb_client = chromadb.PersistentClient(path="./chroma_db")
chromadb_collection= chromadb_client.get_or_create_collection(name="default")

def save_embeddings(chunks: list[str],embeddings:list[list[float]])->None:
    ids=[str(i) for i in range(len(chunks))]
    chromadb_collection.add(
        documents=chunks,
        embeddings=embeddings,
        ids=ids
    )

save_embeddings(chunks,embeddings)
```

```python
def retrieve(query: str, top_k: int = 20) -> list[dict]:
    query_vector = get_embedding(query, is_query=True)

    results = chromadb_collection.query(
        query_embeddings=[query_vector],
        n_results=top_k,
        include=["documents", "metadatas", "distances"] # 要求回傳更多資訊
    )

    formatted_results = []
    if results['documents']:
        for i in range(len(results['documents'][0])):
            formatted_results.append({
                "content": results['documents'][0][i],
                "metadata": results['metadatas'][0][i] if results['metadatas'] else {},
                "score": results['distances'][0][i] # 距離越小越相似
            })
    
    return formatted_results

#測試query
query="「極光專案」中提到的「代號：貝殼 (Project Shell)」其具體操作金額是多少？"
retrieved_chunks=retrieve(query,10)

for i ,chunk in enumerate(retrieved_chunks):
    print(f"[{i}] {chunk}\n")

print(f"召回了 {len(retrieved_chunks)} 筆候選資料，")
```

```python
from sentence_transformers import CrossEncoder
import warnings

warnings.filterwarnings('ignore')

cross_encoder = CrossEncoder('BAAI/bge-reranker-v2-m3')

def rerank_with_scores(query: str, retrieved_chunks: list, top_k: int) -> list[tuple[str, float]]:
    # 從字典中提取 content 欄位
    text_chunks = []
    for chunk in retrieved_chunks:
        if isinstance(chunk, dict) and 'content' in chunk:  # 如果是字典且有 content 鍵
            text_chunks.append(chunk['content'])
        elif isinstance(chunk, str):
            text_chunks.append(chunk)
        elif hasattr(chunk, 'node'):
            text_chunks.append(chunk.node.text)
        elif hasattr(chunk, 'text'):
            text_chunks.append(chunk.text)
        else:
            text_chunks.append(str(chunk))
    
    if not text_chunks:
        return []
    
    top_k = min(top_k, len(text_chunks))
    
    pairs = [(query, chunk) for chunk in text_chunks]
    scores = cross_encoder.predict(pairs)
    scored_chunks = list(zip(text_chunks, scores))
    scored_chunks.sort(key=lambda x: x[1], reverse=True)
    
    return scored_chunks[:top_k]

# 執行重排並顯示分數
reranked_results = rerank_with_scores(query, retrieved_chunks, 3)

print("=" * 80)
print(f"查詢問題: {query}")
print("=" * 80)
print("\n重排後的結果 (Top 3):\n")

for i, (chunk, score) in enumerate(reranked_results, 1):
    print(f"[排名 {i}] 相關性分數: {score:.4f}")
    print("-" * 80)
    print(chunk)
    print("-" * 80)
    print()
```
