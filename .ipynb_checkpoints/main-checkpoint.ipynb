```python
from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

loader = TextLoader("doc.md", encoding="utf-8")
data = loader.load()

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size = 300,        # 每個片段大約 300 字
    chunk_overlap  = 30,     # 片段間重疊 30 字
    add_start_index = True,  # 記住每個片段在原文中的位置
)
recursive_splits = text_splitter.split_documents(data) 
print(f"原本 1 個文檔被切成了 {len(recursive_splits)} 個片段")
print(f"第一個片段內容：\n{recursive_splits[0].page_content}")
```

```python
from sentence_transformers import SentenceTransformer
embedding_model = SentenceTransformer("Qwen/Qwen3-Embedding-0.6B", trust_remote_code=True)
def get_embedding(text, is_query=False):
    prefix = "query: " if is_query else "passage: "
    return embedding_model.encode([prefix + text])[0].tolist()

test_text = "test"
vector = get_embedding(test_text)

print(f"資料型別: {type(vector)}")
print(f"向量長度: {len(vector)}")
print(f"前五個數值: {vector[:5]}")
```
